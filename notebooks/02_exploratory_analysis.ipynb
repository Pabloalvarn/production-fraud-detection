# 02 — Exploratory analysis
Focus on rare-event prevalence, time, transaction type, amount and feature availability. Every chart should lead to a written conclusion.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from fraud_detection.config import load_config, project_path

config = load_config("configs/base.yaml")
df = pd.read_csv(project_path(config["paths"]["raw_data"]))
```

```python
prevalence = df["isFraud"].mean()
print(f"Fraud prevalence: {prevalence:.4%}")
df.groupby("type")["isFraud"].agg(["count", "sum", "mean"]).sort_values("mean", ascending=False)
```

```python
temporal = df.groupby("step")["isFraud"].agg(transactions="count", frauds="sum", prevalence="mean")
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
temporal["transactions"].plot(ax=axes[0], title="Transaction volume by step")
temporal["prevalence"].plot(ax=axes[1], title="Fraud prevalence by step")
plt.tight_layout()
```

```python
plot_frame = df.assign(log_amount=np.log1p(df["amount"])).sample(
    min(200_000, len(df)), random_state=42
)
sns.histplot(
    data=plot_frame, x="log_amount", hue="isFraud", stat="density", common_norm=False, bins=60
)
plt.title("Log transaction amount by class")
```
