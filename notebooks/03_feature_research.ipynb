# 03 — Feature research
Investigate leakage-safe features. The production implementation belongs in `src/fraud_detection/features/`, not in this notebook.

```python
import pandas as pd

from fraud_detection.config import load_config, project_path
from fraud_detection.features.build_features import build_online_features

config = load_config("configs/base.yaml")
raw = pd.read_csv(project_path(config["paths"]["raw_data"]))
features = build_online_features(raw)
features.head()
```

```python
numeric = features.select_dtypes("number").join(raw["isFraud"])
numeric.groupby("isFraud").median().T.rename(columns={0: "non_fraud_median", 1: "fraud_median"})
```

## Feature acceptance checklist
- Is the value available before the decision?
- Is it computed using past observations only?
- Can the same transformation run in the API?
- Is division by zero handled?
- Is the feature stable over time?
- Does it encode another fraud rule or the label?

## Advanced extension
