# 04 — Final model analysis
Run training through the package CLI first. This notebook reads locked outputs and creates communication-ready analysis; it must not tune the model on test results.

```python
import json

import matplotlib.pyplot as plt
import pandas as pd

from fraud_detection.config import project_path

metrics = json.loads(project_path("reports/metrics/test_metrics.json").read_text())
predictions = pd.read_csv(project_path("reports/metrics/test_predictions.csv"))
pd.Series(metrics, name="value").to_frame()
```

```python
predictions.groupby("step").agg(
    events=("y_true", "size"), frauds=("y_true", "sum"), mean_score=("probability", "mean")
).plot(subplots=True, figsize=(12, 8))
plt.tight_layout()
```
